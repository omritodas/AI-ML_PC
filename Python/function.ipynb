## Peramitter

```python
#Basic

def hello(ok):

    print(f"omrito {ok}")

hello("user")

#Multiple

def mul(a,b):

    ans=a*b

    print(ans)

mul(10,20)

#Return

def Squer(a):

    ans=a**2

    return ans

ans=Squer(10)+5

print(ans)

#Variable lenth Peramitter

def squer_sum(*args):

    sum=0

    for i in args:

        sum=sum+i**2

    return sum

print(squer_sum(1,2,3,4,5))

#Keyword Peramitter

def user(**keyargs):

    for key,val in keyargs.items():

        print(f"{key} : {val}")

user(name="omrito",id=1008,cls=15)
```
